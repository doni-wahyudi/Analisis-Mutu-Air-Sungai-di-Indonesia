```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("status_mutu_air.csv")

# Step 1: Clean classification values (lowercase + remove spaces)
df.iloc[:, 3:] = df.iloc[:, 3:].applymap(
    lambda x: x.lower().replace(" ", "") if isinstance(x, str) else x
)

# Step 2: Fill missing values horizontally
df.iloc[:, 3:] = df.iloc[:, 3:].ffill(axis=1)
df.iloc[:, 3:] = df.iloc[:, 3:].bfill(axis=1)

# Step 3: Define mapping from text to numbers
class_mapping = {
    'memenuhi': 0,
    'memenuhicemarringan': 5,
    'cemarringan': 10,
    'cemarringancemarringan': 10,
    'cemarringancemarsedang': 15,
    'cemarringansedang': 15,
    'cemarringancemarberat': 18,
    'cemarsedang': 20,
    'cemarsedangcemarberat': 25,
    'cemarsedangberat': 25,
    'memenuhicemarringancemarberat': 25,
    'cemarberat': 30
}

# Step 4: Apply mapping
df_numeric = df.copy()
df_numeric.iloc[:, 3:] = df_numeric.iloc[:, 3:].replace(class_mapping)

# Step 5: Group data by province
province_groups = df_numeric.groupby("provinsi")

# Step 6: Plot each province separately
for provinsi, group_df in province_groups:
    plt.figure(figsize=(10, 6))
    tahun = [2017, 2018, 2019, 2020, 2021, 2022]

    for _, row in group_df.iterrows():
        sungai = row['Sungai']
        mutu_list = [row[str(t)] for t in tahun]
        plt.plot(tahun, mutu_list, marker='o', label=sungai)

    plt.xlabel('Tahun')
    plt.ylabel('Tingkat Pencemaran')
    plt.title(f'Tren Mutu Air Sungai di {provinsi} (2017–2022)')
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# Step 1: Load data
df = pd.read_csv("status_mutu_air.csv")

# Step 2: Clean classification values (lowercase + remove spaces)
df.iloc[:, 3:] = df.iloc[:, 3:].applymap(
    lambda x: x.lower().replace(" ", "") if isinstance(x, str) else x
)

# Step 3: Fill missing values horizontally
df.iloc[:, 3:] = df.iloc[:, 3:].ffill(axis=1)
df.iloc[:, 3:] = df.iloc[:, 3:].bfill(axis=1)

# Step 4: Classification to numeric score
class_mapping = {
    'memenuhi': 0,
    'memenuhicemarringan': 5,
    'cemarringan': 10,
    'cemarringancemarringan': 10,
    'cemarringancemarsedang': 15,
    'cemarringansedang': 15,
    'cemarringancemarberat': 18,
    'cemarsedang': 20,
    'cemarsedangcemarberat': 25,
    'cemarsedangberat': 25,
    'memenuhicemarringancemarberat': 25,
    'cemarberat': 30
}

df_numeric = df.copy()
df_numeric.iloc[:, 3:] = df_numeric.iloc[:, 3:].replace(class_mapping)

# Step 5: Define interval bands (level categories)
pollution_bands = [
    (0, 10, 'Baik', '#c6f5d5'),
    (10, 25, 'Sedang', '#ffe599'),
    (25, 35, 'Buruk', '#f4cccc')
]

# Step 6: Group by province and plot
province_groups = df_numeric.groupby("provinsi")
tahun = [2017, 2018, 2019, 2020, 2021, 2022]

for provinsi, group_df in province_groups:
    plt.figure(figsize=(10, 6))
    added_bands = set()

    # Add colored pollution level background
    for low, high, label, color in pollution_bands:
        plt.axhspan(low, high, facecolor=color, alpha=0.4, label=label)

    # Plot river trends
    for _, row in group_df.iterrows():
        sungai = row['Sungai']
        mutu_list = [row[str(t)] for t in tahun]
        plt.plot(tahun, mutu_list, marker='o', label=sungai)

    # Custom legend (bands + river lines)
    handles, labels = plt.gca().get_legend_handles_labels()
    band_patches = [mpatches.Patch(color=color, alpha=0.4, label=label) for _, _, label, color in pollution_bands]

    plt.xlabel('Tahun')
    plt.ylabel('Tingkat Pencemaran')
    plt.title(f'Tren Mutu Air Sungai di {provinsi} (2017–2022)')
    plt.legend(handles=band_patches + handles, bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import math

# ✅ Step 1: Create the mutu_dict (unchanged)
mutu_dict = {}

for index, row in df_numeric.iterrows():
    provinsi = row['provinsi']
    sungai = row['Sungai']
    mutu_list = [row['2017'], row['2018'], row['2019'], row['2020'], row['2021'], row['2022']]
    if provinsi in mutu_dict:
        mutu_dict[provinsi][sungai] = mutu_list
    else:
        mutu_dict[provinsi] = {sungai: mutu_list}

# ✅ Step 2: Plot all provinces using subplots
def plot_mutu_air_subplot(mutu_dict):
    tahun = [2017, 2018, 2019, 2020, 2021, 2022]
    provinces = list(mutu_dict.keys())
    n = len(provinces)

    cols = 3
    rows = math.ceil(n / cols)

    fig, axes = plt.subplots(rows, cols, figsize=(cols * 5, rows * 4), constrained_layout=True)
    axes = axes.flatten()

    for i, provinsi in enumerate(provinces):
        ax = axes[i]
        sungai_dict = mutu_dict[provinsi]
        for sungai, mutu_list in sungai_dict.items():
            ax.plot(tahun, mutu_list, marker='o', label=sungai)

        ax.set_title(provinsi)
        ax.set_xlabel('Tahun')
        ax.set_ylabel('Tingkat Pencemaran')

        # Add interval background
        ax.axhspan(0, 11, facecolor='#c6f5d5', alpha=0.4, label='Baik' if i == 0 else None)
        ax.axhspan(11, 29, facecolor='#ffe599', alpha=0.4, label='Sedang' if i == 0 else None)
        ax.axhspan(29, 35, facecolor='#f4cccc', alpha=0.4, label='Buruk' if i == 0 else None)

        ax.legend(fontsize='x-small', loc='upper right')

    # Hide extra subplots
    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])

    plt.suptitle("Tren Mutu Air Sungai per Provinsi (2017–2022)", fontsize=16)
    plt.show()

# ✅ Call the function
plot_mutu_air_subplot(mutu_dict)
```

```python
import matplotlib.pyplot as plt
import math

# Step 1: keep your original mutu_dict logic
mutu_dict = {}

for index, row in df_numeric.iterrows():
    provinsi = row['provinsi']
    sungai = row['Sungai']
    mutu_list = [row['2017'], row['2018'], row['2019'], row['2020'], row['2021'], row['2022']]
    if provinsi in mutu_dict:
        mutu_dict[provinsi][sungai] = mutu_list
    else:
        mutu_dict[provinsi] = {sungai: mutu_list}

# Step 2: Subplot per province, 1 subplot per sungai
def plot_per_province(mutu_dict):
    tahun = [2017, 2018, 2019, 2020, 2021, 2022]

    for provinsi, sungai_dict in mutu_dict.items():
        n = len(sungai_dict)
        cols = 2
        rows = math.ceil(n / cols)

        fig, axes = plt.subplots(rows, cols, figsize=(cols * 5, rows * 4), constrained_layout=True)

        # ✅ Fix axes shape
        if isinstance(axes, plt.Axes):
            axes = [axes]
        else:
            axes = axes.flatten()

        for i, (sungai, mutu_list) in enumerate(sungai_dict.items()):
            ax = axes[i]
            ax.plot(tahun, mutu_list, marker='o', label=sungai)

            ax.set_title(sungai)
            ax.set_ylim(-2, 35)
            ax.set_xlabel('Tahun')
            ax.set_ylabel('Tingkat Pencemaran')

            # Interval bands
            ax.axhspan(-2, 11, facecolor='#c6f5d5', alpha=0.4)
            ax.axhspan(11, 29, facecolor='#ffe599', alpha=0.4)
            ax.axhspan(29, 35, facecolor='#f4cccc', alpha=0.4)

            ax.grid(True, linestyle='--', alpha=0.5)

        # Hapus subplot kosong
        for j in range(i + 1, len(axes)):
            fig.delaxes(axes[j])

        fig.suptitle(f'Tren Mutu Air Sungai di Provinsi {provinsi} (2017–2022)', fontsize=14)
        plt.show()

# Run
plot_per_province(mutu_dict)
```
